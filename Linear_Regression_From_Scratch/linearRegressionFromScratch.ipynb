```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("study_scores_noisy_100.csv")
data.head()
```

```python
def loss_function(m,b,points):
  total_error = 0
  for i in range(len(points)):
    x = points.iloc[i].Hours_Studied
    y = points.iloc[i].Exam_Score
    total_error += (y-(m*x+b))**2
  total_error/float(len(points))
def gradient_descent(m_now,b_now,points,learning_rate):
  m_gradient = 0
  b_gradient = 0
  n = len(points)
  for i in range(n):
    x = points.iloc[i].Hours_Studied
    y = points.iloc[i].Exam_Score
    m_gradient +=-(2/n)*x*(y-(m_now*x+b_now))
    b_gradient +=-(2/n)*(y-(m_now*x+b_now))
  m=m_now-m_gradient*learning_rate
  b=b_now-b_gradient*learning_rate
  return m,b
m=0
b=0
learning_rate = 0.0001
epochs = 1000
for i in range(epochs):
  if i%100==0:
     print(f"epochs {i}")
  m,b = gradient_descent(m,b,data,learning_rate)
print(m,b)
plt.scatter(data.Hours_Studied,data.Exam_Score,color = 'red')
plt.plot(data.Hours_Studied,m*data.Hours_Studied+b,color = 'blue')
plt.show()
```
